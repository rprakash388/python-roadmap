## How do we convert and transform the column data

### 1. One Hot Encoder

This can be achieved by using any of the following 2 libraries:

1. Pandas - pd.getdummies() - it is used to creating the unique column values as new columns
2. sklearn - OneHotEncoder()

#### 1.1 Pandas - pd.getdummies()

```python
weather_data_copy_02 = weather_data.copy()

weather_data_copy_02
```

```python
weather_data_copy_02["Weather"].unique()
```

```python
# pd.getdummies() - it is used to creating the unique column values as new columns
weather_data_copy_02 = pd.get_dummies(data=weather_data_copy_02, columns=["Weather"], dtype="int")

weather_data_copy_02
```

```python
weather_data_copy_02.dtypes
```

```python
weather_data_copy_03 = weather_data.copy()

weather_data_copy_03
```

### 2. sklearn - OneHotEncoder()

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# 1. Make a copy of your original DataFrame
weather_data_copy_03 = weather_data.copy()

# 2. Check the DataFrame
print(weather_data_copy_03.head())

# 3. Create OneHotEncoder
ohe = OneHotEncoder(sparse_output=False)

# 4. Encode the Weather column
weather_encoded = ohe.fit_transform(
    weather_data_copy_03[["Weather"]]
)

# 5. Display encoded values
print(weather_encoded)

# 6. Get the new column names
encoded_columns = ohe.get_feature_names_out(["Weather"])

print(encoded_columns)

# 7. Convert encoded data into a DataFrame
encoded_df = pd.DataFrame(
    weather_encoded,
    columns=encoded_columns,
    index=weather_data_copy_03.index
)

# 8. Add encoded columns to the original DataFrame
weather_data_copy_03 = pd.concat(
    [weather_data_copy_03, encoded_df],
    axis=1
)

# 9. Display the final DataFrame
print(weather_data_copy_03.head())
```

## How to choose between Label Encoder and One Hot Encoder?

### Input feature:
Based on whether it is **Parametric or a Non Parametric Model**, we will choosing between LE or OHE.
* For **Parametric Models** - better to go with **OHE.**
* For **Non-Parametric Models** - always go with **Label Encoding.**

### Output feature:
Always the output feature, it must be **Label Encoder.**
