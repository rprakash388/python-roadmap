## Chapter #1 - Machine Learning Overview

### 1. What is Machine Learning?

 - Insights + Prediction
 - Machine Learning means teaching a computer using examples (data), so that it can learn patterns and give answers for new data.

**E.g**
 - **YouTube** - ML can recommend videos based on what you watch.
 - **Amazon / Shopping websites** - ML can recommend products
 - **Gmail** - ML helps identify.

### 2. Why Machine Learning required?

 - Machine Learning is required because traditional programming cannot easily handle every problem using fixed rules. ML allows computers to learn from large amounts of data, find patterns, and make predictions or decisions.

E.g
**Problem Statement:** Manual rules break down in complex problems

**Example:** Imagine you want to detect spam emails.

**In Traditional Ways:** We can defined the rules based on email content. It can't be handled manual for all senarioes

**Solution:** Build a system that automatically does this job and scale it.

**Beneifts:** Saves cost, improves decisions

Previous Emails > Machine Learning > Learns patterns > New Email > Spam / Not Spam




**5 Reason Main Reasons:**
1. To handle large amounts of data
2. To make predictions
3. To automate tasks
4. To find hidden patterns
5. To improve with experience




### 3. Difference between Machine Learning and Statistics?

 - Statistics mainly focuses on understanding and explaining data.
 - Machine Learning mainly focuses on making predictions or decisions from data.



### 4. Types of Machine Learning?

1. **Supervised** Learning - X and Y present in the data
2. **Unsupervised** Learning - Only X present
3. **Reinforcement** Learning - Like training a kid with rewards system
