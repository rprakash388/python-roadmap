## 4. Data Preparation

```python
cars_data
```

```python
x = cars_data[["HP", "VOL", "SP", "WT"]]

y = cars_data["MPG"]

y
```

### There is nothing to prepare because all features are numeric values

## 5. Model Building

```python
linear_model = LinearRegression()
```

## 6. Model Training

```python
linear_model.fit(x,y)
```

```python
linear_model.intercept_
```

```python
linear_model.coef_
```

## 7. Model Testing

```python
y_pred = linear_model.predict(x)

y_pred
```

## 8. Model Evaluation

```python
error = y-y_pred

error
```

```python
error.mean()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
```

```python
mean_absolute_error(y, y_pred) # same units
```

```python
mean_squared_error(y, y_pred) # squared output units
```

```python
root_mean_squared_error(y, y_pred) # normalize units
```

```python
r2_score(y, y_pred)
```

```python
import statsmodels.formula.api as smf
```

### To see Adjusted R2 Score, we are using OLS technique from StatsModel Formula

```python
linear_model = smf.ols(formula="MPG~HP", data=cars_data).fit()
```

```python
print("R2 Score:", linear_model.rsquared.round(4), "\nR2 Adjusted Score:", linear_model.rsquared_adj.round(4))
```

```python
linear_model = smf.ols(formula="MPG~HP+VOL", data=cars_data).fit()
```

```python
print("R2 Score:", linear_model.rsquared.round(4), "\nR2 Adjusted Score:", linear_model.rsquared_adj.round(4))
```

```python
linear_model = smf.ols(formula="MPG~HP+VOL+SP", data=cars_data).fit()
```

```python
print("R2 Score:", linear_model.rsquared.round(4), "\nR2 Adjusted Score:", linear_model.rsquared_adj.round(4))
```

```python
linear_model = smf.ols(formula="MPG~HP+VOL+SP+WT", data=cars_data).fit()
```

```python
print("R2 Score:", linear_model.rsquared.round(4), "\nR2 Adjusted Score:", linear_model.rsquared_adj.round(4))
```
