## 1. Groupby function

```python
import pandas as pd
```

```python
insurance_data = pd.read_csv("Insurance.csv")
insurance_data
```

### 1.1 Perform Initial Analysis

```python
insurance_data.isna().sum()
```

```python
insurance_data.dtypes
```

```python
insurance_data.describe()
```

### 1.2 Pull up the average Insurance charges based on region

```python
insurance_data["region"].unique()
```

```python
insurance_data.groupby(by="region")["charges"].mean().sort_values().round(2)
```

### 1.2 Pull up the average Insurance charges based on region and sex

```python
insurance_data.groupby(by=["region", "sex"])["charges"].max().sort_values(ascending=False).round(2)
```

## 2. Pivot Table

```python
pd.pivot_table(data=insurance_data, values="charges", index="region", columns=["sex","children"]).round()
```

## 3. Aggregate

### 3.1 Find out the average insurance charges based on Region

```python
insurance_data.groupby(by=["region"])["charges"].mean().sort_values(ascending=False).round(2)
```

```python
insurance_data.groupby(by=["region","sex"])["charges"].agg(func=["mean", "min","max","sum"]).round()
```

```python
sales_all_data = pd.concat(objs=[sales_data_2017,sales_data_2018, sales_data_2019])
sales_all_data
```

```python
sales_all_data.dtypes
```

```python
sales_all_data.isna().sum()
```

```python
sales_all_data.drop_duplicates(inplace=True)
sales_all_data
```

```python
sales_all_data.isna().sum()
```

```python
valid_sales_all_data = sales_all_data[sales_all_data["Date"].notna() & 
    sales_all_data["Voucher"].notna()  & 
    #sales_all_data["Party"].notna() &
    #sales_all_data["Product"].notna()  & 
    sales_all_data["Qty"].notna()  & 
    sales_all_data["Rate"].notna()]

valid_sales_all_data
```

```python
valid_sales_all_data.isna().sum()
```

```python
# valid_sales_all_data = valid_sales_all_data.drop(columns=["Gross", "Disc", "Voucher Amount"])
valid_sales_all_data = valid_sales_all_data[["Date","Voucher","Party","Product","Qty","Rate"]]
valid_sales_all_data
```

```python
valid_sales_all_data.isna().sum()

valid_sales_all_data.shape
```

```python
valid_sales_all_data.dtypes
```

```python
valid_sales_all_data = valid_sales_all_data.copy()

valid_sales_all_data["Voucher"] = (
    valid_sales_all_data["Voucher"]
    .astype(str)
    .str.replace("Sal:", "", regex=False)
)

valid_sales_all_data
```

```python
valid_sales_all_data.isna().sum()
```

```python
valid_sales_all_data.dtypes
```

```python
valid_sales_all_data = valid_sales_all_data.replace(r'^\s*$', pd.NA, regex=True)
valid_sales_all_data = valid_sales_all_data.dropna().reset_index(drop=True)

valid_sales_all_data
```

```python
valid_sales_all_data["Voucher"] = valid_sales_all_data["Voucher"].str.replace("Sal:","").astype(int)
```

```python
valid_sales_all_data.dtypes
```

```python
valid_sales_all_data["Qty"] = pd.to_numeric(
    valid_sales_all_data["Qty"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
).astype(int)
```

```python
valid_sales_all_data.dtypes
```

```python
valid_sales_all_data
```

```python
valid_sales_all_data["Rate"] = pd.to_numeric(
    valid_sales_all_data["Rate"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
).astype(float)
```

```python
valid_sales_all_data.dtypes
```

```python
valid_sales_all_data
```

```python
valid_sales_all_data.isna().sum()
```

```python
valid_sales_all_data["Date"] = pd.to_datetime(
    valid_sales_all_data["Date"],
    format="%d/%m/%Y",
    errors="coerce"
)
```

## 6. Merge

**Common merge types**

**left** — keep all rows from the first DataFrame

**right** — keep all rows from the second DataFrame

**inner** — keep only matching rows

**outer** — keep all rows from both

```python
merge_data = pd.merge(
    sales_data_2017, 
    sales_data_2018,
    how="inner",
left_on="Date",
right_on="Date")

merge_data
```

##### 7.1 Create a new column in the same dataframe with the name "Eligibility Criteria" where the age > 20, say Eligible else Not Eligible

```python
def get_eligible_by_age(x):
    if x > 20:
        return "Eligible"
    else:
        return "Not Eligible"
```

```python
insurance_data["Eligibility Criteria"] = insurance_data["age"].apply(get_eligible_by_age)

insurance_data
```

### 8. Lambda function

* Anoynous / nameless function

```python
insurance_data
```

```python
insurance_data["EC"] = insurance_data["age"].apply(lambda x: "Eligible" if x > 20 else "Not Eligible")

insurance_data
```
