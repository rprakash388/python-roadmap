## Task #2 - Merge Function 1 Example

- **merge()** is used to combine two DataFrames based on a common column (key), similar to SQL JOIN operations.

```python
employee_data = pd.read_csv("employees.csv")

employee_data
```

```python
department_data = pd.read_csv("departments.csv")

department_data
```

### 2.1 Inner Join

- Returns only matching records from both DataFrames.

```python
result = pd.merge(
        employee_data, 
        department_data, 
        on="DEPARTMENT_ID", 
        how="inner",
        suffixes=("_Employee", "_Department") # If tables has same column name then it will append suffixes
    )

result
```

#### 2.1.1 - Select specific columns

```python
result = pd.merge(
        employee_data, 
        department_data, 
        on="DEPARTMENT_ID", 
        how="inner"
    )[["EMPLOYEE_ID", "FIRST_NAME", "LAST_NAME", "PHONE_NUMBER", "SALARY", "DEPARTMENT_NAME", "DEPARTMENT_ID"]]

result
```

```python
result = pd.merge(
        employee_data, 
        department_data, 
        on="DEPARTMENT_ID", 
        how="inner"
    )[["EMPLOYEE_ID", "FIRST_NAME", "LAST_NAME", "PHONE_NUMBER", "SALARY", "DEPARTMENT_NAME", "DEPARTMENT_ID"]]

result.columns = ["EMPLOYEE ID", "FIRST NAME", "LAST NAME", "PHONE NUMBER", "SALARY", "DEPARTMENT NAME", "DEPARTMENT ID"] # Rename the columns

result
```

### 2.2 Left Join

 - Returns all records from the left DataFrame (Employee).

```python
result = pd.merge(
        employee_data, 
        department_data, 
        on="DEPARTMENT_ID", 
        how="left"
    )[["EMPLOYEE_ID", "FIRST_NAME", "LAST_NAME", "PHONE_NUMBER", "SALARY", "DEPARTMENT_NAME", "DEPARTMENT_ID"]]

result
```

### 2.3 Right Join

 - Returns all records from the right DataFrame (Department) with matching employee record details. If not matching then it should shows 'NaN'

```python
result = pd.merge(
        employee_data, 
        department_data, 
        on="DEPARTMENT_ID", 
        how="right"
    )[["EMPLOYEE_ID", "FIRST_NAME", "LAST_NAME", "PHONE_NUMBER", "SALARY", "DEPARTMENT_NAME", "DEPARTMENT_ID"]]

result
```

### 2.4 Outer Join

- Returns all records from both DataFrames

```python
result = pd.merge(
        employee_data, 
        department_data, 
        on="DEPARTMENT_ID", 
        how="outer"
    )[["EMPLOYEE_ID", "FIRST_NAME", "LAST_NAME", "PHONE_NUMBER", "SALARY", "DEPARTMENT_NAME", "DEPARTMENT_ID"]]

result
```

## Task #3 - List Comprehension 2 examples

 - List comprehension is a short way to create a new list from an existing iterable

#### 3.1 Get even numbers

```python
numbers = [1, 2, 3, 4, 5, 6]

even_numbers = [x for x in numbers if x % 2 == 0]

print(even_numbers)
```

#### 3.2 Multiple if conditions

```python
numbers = [2, 4, 6, 8, 10, 12, 15]

result = [x for x in numbers if x % 2 == 0 if x > 5]

print(result)
```

#### 3.3 Multiple for loops

```python
numbers = [1, 2, 3]
letters = ['A', 'B']

result = [(x, y) for x in numbers for y in letters]

print(result)
```

#### 3.4 Multiple for + if

```python
result = [
    (x, y)
    for x in range(1, 4)
    for y in range(1, 4)
    if x != y
]

print(result)
```

## Task #4 Lambda Function 2 examples with Map, Filter, Reduce

- A lambda function is an anonymous function written in a single line.

### 4.1 Map

 - Transforms each element in a collection and returns a new collection containing the transformed values.

#### 4.1.1 Multiply every number by 2

```python
numbers = [2,4,7,19,45,50, 34,67,98]

result = list(map(lambda x: x * 2, numbers))

result
```

#### 4.1.2 Convert names to Uppercase

```python
names = ["Prakash", "Raju", "Kamala", "Sharmila", "Divyadharshana", "Gokulakannan"]

result = list(map(lambda x: x.upper(), names))

result
```

### 4.2 Filter

 - Selects elements from a collection that satisfy a given condition.

#### 4.2.1 Get only even numbers

```python
numbers = [2,4,7,19,45,50, 34,67,98]

result = list(filter(lambda x : x % 2 == 0, numbers))

result
```

#### 4.2.2 Get the above 30 numbers

```python
numbers = [2,4,7,19,45,50, 34,67,98]

result = list(filter(lambda x : x  > 30, numbers))

result
```

### 4.3 Reduce

 - Keeps only elements where the condition is True.
 - Combines all elements of a collection into a single value using an accumulator.

```python
from functools import reduce
```

#### 4.3.1 Calculate the sum of all numbers

```python
numbers = [30000, 45600, 5432.56, 34547]

result = reduce(lambda x, y: x + y, numbers)

result
```

#### 4.3.2 Find the largest number

```python
numbers = [30000, 45600, 5432.56, 34547]

result = reduce(lambda x, y:x if x > y else y, numbers)

result
```
