### Number Game
1. Instructions
2. Set the secret number
3. Read the input from User
4. Implement the logic
5. Results

```python
import random as rd

secret_number = rd.randint(1, 100)

while True:
    try:
        user_input = int(input("Please enter the valid number from 1 to 100"))
        
        if secret_number == user_input:
            print("Congrats!!! You're guess is right!")
            break
        elif secret_number > user_input:
            print("The given number is greater than secret number")
        elif secret_number < user_input:
            print("The given number is lesser than secret number")
        else:
            print("Sorry. Your guess is wrong")

    except:
        print("Please enter only number from 1 to 100")

```
