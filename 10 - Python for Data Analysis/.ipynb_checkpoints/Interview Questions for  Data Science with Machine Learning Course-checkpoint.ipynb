#### 3.	When would you use a list over a set or a tuple?

Use a list when you need an ordered and changeable collection.

**List vs Set vs Tuple**
- List → Use when you need ordered data and want to change the values.

Example: 
       ["Apple", "Banana", "Mango"]

 - Set → Use when you need unique values and don't care about order.

      Example: {"Apple", "Banana", "Mango"}

- Tuple → Use when you need ordered data that should not change.

      Example: ("Apple", "Banana", "Mango")

#### 9.	When should you use .loc[] and when .iloc[]?

**.loc[] vs .iloc[]**
 - .loc[] → Used to select data using labels/names.
 - .iloc[] → Used to select data using index positions/numbers.

**Simple answer:**

👉 .loc[] = Labels

👉 .iloc[] = Positions


```python
df.loc[5, "Salary"]     # Using row label and column name
df.iloc[5, 2]           # Using row position and column position
```

#### 3.	Write a program that handles division by zero using try-except

```python
try:
    numerator = int(input("Enter Numerator number"))
    denomenator = int(input("Enter Denomenator number"))

    result = numerator / denomenator 

    print(result)
    
except ValueError:
    print("Please enter only numeric values only")
    
except ZeroDivisionError:
    print("Cannot divide by zero")
```

#### 4.	Given a string, count how many vowels it has.

```python
v_count = 0
vowels = ['a', 'e', 'i', 'o', 'u']

user_input = input("Please enter your string: ").lower()

for c in user_input:
    if c in vowels:
        v_count += 1

print(f"The vowels count is: {v_count}")
```

#### 5.	Reverse a list without using built-in functions.

```python
list_data = ["Prakash", "Raju", "Sharmila", "Kamala", "Divyadharshana", "Gokulakannan"]

rev_list = []

for name in list_data[::-1]:
    rev_list.append(name)

print(rev_list)
```

```python
rev_list = []

for name in list_data[-1:-(len(list_data)+1):-1]:
    rev_list.append(name)

print(rev_list)
```

```python
rev_list = []

for i in range(len(list_data) - 1, -1,-1):
    rev_list.append(list_data[i])

print(rev_list)
```
