## 2. Numpy - Numerical Operations

```python
import numpy as np
```

```python
np.repeat(a=data, repeats=2, axis=1) # Duplicates the column values with x times
```

```python
np.repeat(a=data, repeats=2, axis=0)  # Duplicates the row values with x times
```

```python
newDataFrame = pd.DataFrame(data=np.repeat(a=data, repeats=2, axis=0))
newDataFrame
```
