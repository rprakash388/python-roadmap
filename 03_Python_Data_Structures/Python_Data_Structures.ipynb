## 1. List
* 1-Dimensional
* Allow duplicate values
* Hytrogenious
* Indexing and Slicing is applicable
* Muttable

```python
list_data = []
```

```python
type(list_data)
```

```python
list_data=[45,67,85,34]
```

```python
list_data
```

```python
list_data.append("Apple")
list_data
```

```python
list_data[5]="Orange"
list_data
```

```python
list_data[0:2]
```

```python
list_data[-2:-1]
```

## 2. Sets
* 1-Dimensional
* Allow unique values
* Heterogeneous Data Types
* Indexing and Slicing is not applicable
* Muttable

```python
set_data=set()
set_data
```

```python
type(set_data)
```

```python
set_data.add("Apple")
set_data
```

```python
set_data.add("Orange")
set_data.add("Banana")
set_data.add(34)
```

```python
set_data
```

```python
set_data.discard(34)
set_data
```

## 3. Tuples
* 1-Dimesional
* Allow duplicate values
* Heterogeneous Data Types
* Indexing and Slicing is applicable
* Immuttable

```python
tuple_data=()
tuple_data
```

```python
type(tuple_data)
```

```python
tuple_data=(23,56,78,"Apple", "Orange", "Apple")
tuple_data
```

```python
tuple_data[0]
```

```python
tuple_data[2]="Banana"
```

```python
tuple_data.index(78)
```

## 4. Dictionary
* 2-Dimensional
* Allow duplicate key and value pairs but updating the values if key is duplicate
* Heterogeneous Data Types
* Indexing and Slicing is not applicable
* Muttable

```python
dict_data={"name": "Prakash", "age": 43, "dob": "12/12/1980"}
dict_data
```

```python
dict_data.keys()
```

```python
dict_data.values()
```

```python
dict_data={"name": "Prakash", "age": 43, "dob": "12/12/1980", "name": "Prakash Raju"}
dict_data
```
