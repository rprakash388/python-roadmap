# Task #2 - Create 3 DataFrames using Dictionaries¶

## 2.1 DataFrame from Dictionary of Lists

 - Converts the employee dictionary into a Pandas DataFrame (table) with automatically generated row indexes

```python
import pandas as pd

employee_data = {
    "Name": ["Prakash", "Raj", "Kamala"],
    "Age": [30, 28, 32],
    "Salary": [56000, 45000, 60000]
}

df_employee = pd.DataFrame(employee_data)

print(df_employee)
```

## 2.2 DataFrame from Dictionary of Dictionaries

 - Converts the nested dictionary into a Pandas DataFrame, where Employee1, Employee2, and Employee3 become columns and Name and Salary become rows.

```python
employee_data = {
    "Employee1": {
        "Name": "Prakash",
        "Salary": 56000
    },
    "Employee2": {
        "Name": "Raj",
        "Salary": 45000
    },
    "Employee3": {
        "Name": "Kamala",
        "Salary": 60000
    }
}

df_employee = pd.DataFrame(employee_data)

print(df_employee)
```

## 2.3 DataFrame from List of Dictionaries

 - Converts the list of dictionaries into a Pandas DataFrame, where each dictionary represents one employee and its keys become the column names.

```python
employee_data = [
    {"Name": "Prakash", "Age": 30, "Salary": 56000},
    {"Name": "Raj", "Age": 28, "Salary": 45000},
    {"Name": "Kamala", "Age": 32, "Salary": 60000}
]

df_employee = pd.DataFrame(employee_data)

print(df_employee)
```
