## 1. List

* A List is an ordered, mutable collection that can store multiple elements, including duplicate values and different data types.

```python
employee_salary = [56000, 45000, 28000, 12000]
employee_salary
```

```python
type(employee_salary)
```

#### 1.1 Append function

* Append object to the end of the list.

```python
employee_salary.append("Prakash")
employee_salary
```

#### 1.2 Count function

* Return number of occurrences of value

```python
employee_salary.count("Prakash")
```

#### 1.3 Extend function

* Extend list by appending elements from the iterable.

```python
employee_salary.extend(["Raju", "Kamala"])
employee_salary
```

#### 1.4 Index function

* Return first index of value.

```python
employee_salary.index("Raju")
```

#### 1.5 Insert function

* Insert object before index.

```python
employee_salary
```

```python
employee_salary.insert(6, "Sharmila")
```

```python
employee_salary
```

#### 1.6 Pop function

* Remove and return item at index (default last)

```python
employee_salary
```

```python
employee_salary.pop()
```

```python
employee_salary
```

```python
employee_salary.pop(6)
```

```python
employee_salary
```

### 1.7 Remove function

* Remove first occurrence of value.

```python
employee_salary.remove("Raju")
employee_salary
```

### 1.8 Reverse function

* Reverse a List

```python
employee_salary.reverse()
employee_salary
```

### 1.9 Sort function

* Sort the list in ascending order and return None.
* Applied only Hemogeneous List

```python

employee_salary.sort()
```

```python
employee_names = ["Prakash", "Raju", "Kamala", "Sharmila", "Divyadharshana", "Gokulakannan"]
```

```python
employee_names.sort()
employee_names
```

### 1.10 Copy function

* Return a shallow copy of the list.

```python
new_employee_names = employee_names.copy()
new_employee_names
```

### 1.11 Clear function

* Remove all items from list.

```python
new_employee_names.clear()
new_employee_names
```

### Difference Between append() and extend()

* append() adds the entire list as a single element.
* extend() adds each element individually to the existing list.

### Difference Between remove() and pop()

* **remove** - Removes the first occurrence of a specified value.
* **pop** - Removes an element based on its index and returns the removed value.

## 2. Sets

* A Set is an unordered, mutable collection of unique elements

```python
student_data = {"Prakash", "Raju", "Kamala"}
```

```python
type(student_data)
```

### 2.1 Add function

* Add an element to a set.

```python
student_data.add("Sharmila")
student_data
```

```python
student_data.add("Gokulakannan")
student_data
```

### 2.2 Copy function

* Return a shallow copy of a set.

```python
new_student_data = student_data.copy()
new_student_data
```

### 2.3 Difference function

* Return the difference of two or more sets as a new set.

```python
student_data1 = {"Raju", 50000, 45000}
```

```python
diff_data = student_data.difference(student_data1)
diff_data
```

### 2.4 Difference_Update function

* Remove all elements of another set from this set.

```python
student_data.difference_update(student_data1)
student_data
```

```python
diff_data
```

### Difference between difference and difference_update functions

* **difference()** returns a new set containing elements that are present in the first set but not the second.
* **difference_update()** performs the same operation but updates the original set in-place.

### 2.5 Discard function

* Remove an element from a set if it is a member.

```python
student_data
```

```python
student_data.discard("Prakash")
student_data
```

### 2.6 Intersection function

* Return the intersection of two sets as a new set.

```python
student_data
```

```python
student_data1
```

```python
new_student_data = student_data.intersection(student_data1)
new_student_data
```

### 2.7 Intersection Update function

* Update a set with the intersection of itself and another.

```python
new_student_data.intersection_update(student_data1)
```

```python
new_student_data
```

### Difference between intersection and intersection_update

* **intersection()** - Creates and returns a new set. The original set remains unchanged.
* **intersection_update()** - Updates the original set and keeps only the common elements.

## 4. Dictionary

* A dictionary (dict) is a mutable collection of key-value pairs

```python
employee = {
    "id": 101,
    "name": "Prakash",
    "salary": 56000
}
```

```python
type(employee)
```

```python
employee
```

### 4.1 Copy function

* D.copy() -> a shallow copy of D
* Creates a shallow copy of a dictionary

```python
employee1 = employee.copy()
employee1
```

### 4.2 FromKeys function 

* Create a new dictionary with keys from iterable and values set to value
* Creates a new dictionary using the given keys, with the same default value for every key.

```python
keys = ["name", "salary"]

new_employee = employee.fromkeys(keys)

new_employee
```

### 4.3 Get function

* Return the value for key if key is in the dictionary, else default.
* It s used to retrieve the value associated with a key without raising an error if the key does not exist.

```python
type(employee)
```

```python
employee
```

```python
employee.get("department")
```

```python
print(employee.get("name"))
```

### 4.4 Items function

* D.items() -> a set-like object providing a view on D's items
* Returns all key-value pairs of a dictionary.

```python
employee.items()
```

### 4.5 Keys function

* D.keys() -> a set-like object providing a view on D's keys
* Returns a view containing all the keys in a dictionary.

```python
employee.keys()
```

### 4.6 Pop function

* D.pop(k[,d]) -> v, remove specified key and return the corresponding value.
* Removes a specified key-value pair and returns the value of the removed key.

```python
employee.pop("salary")
```

### 4.7 PopItem function

* Remove and return a (key, value) pair as a 2-tuple.
* Removes and returns the last inserted key-value pair from a dictionary.

```python
employee
```

```python
employee.popitem()
```

### 4.8 SetDefault function

* Insert key with a value of default if key is not in the dictionary.
* Returns the value of a key if the key exists. If the key does not exist, it adds the key with a default value and returns that default value.

```python
employee.setdefault("name", "Raju")
employee
```

### 4.9 Update function

* It is used to add new key-value pairs or update existing key-value pairs in a dictionary.

```python
employee.update({"name": "Prakash", "salary": 56000})
employee
```

### 4.10 Values function

* D.values() -> an object providing a view on D's values
* Returns a view containing all the values in a dictionary.

```python
employee.values()
```
