# Task #4

**Word Guessing Game (Mini Hangman)**

Objective:

Create a word-guessing game where the player tries to guess the letters of a hidden word.

This will test your:

* Loops
* Conditional logic
* String manipulation
* Exception handling
* Logical thinking

Instructions:
1.	Create a list of random words (like “python”, “machine”, “learning”, “bridge”).
2.	Randomly select one word for the game.
3.	Show the word as blanks (_ _ _ _) and let the player guess one letter at a time.
4.	Keep track of wrong guesses (max 6 attempts).
5.	End the game when the player either guesses the full word or uses all chances.


### Updated Code

 - Note: I took some help from ChatGPT

```python
import random

words = ["python", "machine", "learning", "bridge"]

selected_word = random.choice(words)

guessed_word = ["_"] * len(selected_word)

wrong_attempts = 0
max_attempts = 6
guessed_letters = set()

print("===== Word Guessing Game =====")
print(f"The word has {len(selected_word)} letters.")

while wrong_attempts < max_attempts:

    print("\nWord:", " ".join(guessed_word))
    print(f"Wrong attempts: {wrong_attempts}/{max_attempts}")

    try:
        user_input = input("Guess a letter: ").strip().lower()

        if len(user_input) != 1:
            raise ValueError("Please enter only one letter.")

        if not user_input.isalpha():
            raise ValueError("Please enter an alphabetic character.")

        if user_input in guessed_letters:
            print("You already guessed this letter.")
            continue

        guessed_letters.add(user_input)

        if user_input in selected_word:

            print("Correct guess!")

            for index in range(len(selected_word)):
                if selected_word[index] == user_input:
                    guessed_word[index] = user_input

        else:
            wrong_attempts += 1
            print("Wrong guess!")

        if "_" not in guessed_word:
            print("\nCongratulations!")
            print("You guessed the word:", selected_word)
            break

    except ValueError as error:
        print("Invalid input:", error)

else:
    print("\nGame Over!")
    print("The correct word was:", selected_word)
```

## Old Code #2

```python
import random as rd

choices = ['python', 'machine', 'learning', 'bridge']
max_attempts = 10
attempt_count = 0

selected_choice = rd.choice(choices)
fill_word = ['_'] * len(selected_choice)

while attempt_count < max_attempts:

    print("\nWord:", ''.join(fill_word))
    print(f"Please guess the word. It has {len(selected_choice)} characters")

    user_input = input("Enter a character or complete word: ").lower()

    if user_input == selected_choice:
        print("Great!!! Your guess is correct")
        break

    elif len(user_input) == 1:

        if user_input in selected_choice:

            for i, ch in enumerate(selected_choice):
                if ch == user_input:
                    fill_word[i] = user_input

            print("Correct guess!")
        else:
            print("Character not found.")

    else:
        print("Invalid input.")

    attempt_count += 1

    if ''.join(fill_word) == selected_choice:
        print("Great!!! Your guess is correct")
        break

else:
    print(f"Maximum attempts exceeded. The word was '{selected_choice}'")
```
