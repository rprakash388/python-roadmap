## Task #3: Rock, Paper, Scissors Game

**Objective:**

Build a simple Rock-Paper-Scissors game where the player competes with the computer. 

This task will test your understanding of:
* Input/output
* Conditional logic
* Random module
* Loops
* Exception handling

**Instructions:**
1.	The user should enter their choice: rock, paper, or scissors.
2.	The computer should randomly select one of these.
3.	Compare both and print who wins.
4.	Ask if the player wants to play again- if yes, repeat; if no, exit.
5.	Handle invalid inputs gracefully using try-except or condition checks.


```python
import random as rd

choices = ['rock', 'paper', 'scissors']

try:
    continue_to_play = True
    while continue_to_play:
        print("Please enter any one following choice:")
        print("rock")
        print("paper")
        print("scissors")
        input_choice = input("Please enter your choice: ")

        if input_choice not in choices:
            print("Please enter correct choice")
            continue
        
        selected_choice = rd.choice(choices)

        if input_choice == selected_choice:
            print("*** You are the winner ***")
            continue_to_play = False
            while not continue_to_play:
                continue_to_play = input("Do you want to play the game again? press [Y] to Continue, [X] to Exit")
                if continue_to_play.lower() == 'y':
                    continue_to_play = True                       
                elif continue_to_play.lower() == 'x':
                    continue_to_play = False
                    break
                else:
                    print("Invalid Input. Please enter correct input")
                    continue_to_play = False
        else:
            print("*** Sorry, The System is Win ***")
            print("*----------------------------------------------*")        
            print("")
            
except:
    print("Invalid input. Please enter only correct choice")
```
