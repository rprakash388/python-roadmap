## 1. While Loop

### 1.1 Write a Python program using a while loop to print the Fibonacci series of a given number.

```python
a = 0
b = 1
idx = 0
results = []

try:    
    user_input = int(input("Please enter the number: "))
    
    while idx != user_input:
        c = a + b
        results.append(c)
        a = b
        b = c              
        idx += 1    
except:
    print("Invaild input. Please enter only number")

print(results)
```

### 1.2 Write a Python program using a while loop to count how many positive and negative numbers are entered by the user.

```python
pos_count = 0
neg_count = 0
inputs = []
try:
    while True:
        user_input = int(input("Please enter the number: "))

        if user_input == 0:
            break
             
        inputs.append(user_input)
        
        if user_input > 0:
            pos_count += 1
        else:
            neg_count += 1  

    print("The given input numbers are :", inputs)
    
    if pos_count == 0:
        print("No +ve numbers are given")
    else:
        print(f"No. of +ve numbers are {pos_count}") 

    if neg_count == 0:
        print("No -ve numbers are given")
    else:
        print(f"No. of -ve numbers are {neg_count}")     
except:
    print("Invaild input. Please enter only number")
```

## 2. For Loop

### 2.1 Write a Python program using a for loop to count the number of vowels in a given string

```python
vowels = ['a', 'e', 'i', 'o', 'u']
count = 0

for_input = input("Please enter the string")

if len(for_input.split()) > 1:
    print("Invalid input. Please enter only one word string")
else:
    for i in for_input:
        if i.lower() in vowels:
            count += 1

print("Total number of vowels is", count)
```

### 2.2 Write a Python program using a for loop to reverse a given string.

```python
result = ""

for_input = input("Please enter the string")

if len(for_input.split()) > 1:
    print("Invalid input. Please enter only one word string")
else:    
    for i in range(len(for_input)-1, -1, -1):
        result = result + for_input[i]

if len(for_input.split()) == 1:
    print("The reverse string is", result)
```

## 3. If Else

### 3.1 Write a Python program using if-else to check whether a person is eligible to vote based on their age

```python
eligible_vote_age = 18

try:
    input_age = int(input("Please enter your age: "))

    if input_age >= eligible_vote_age:
        print("You are eligible for Vote")
    else:
        print("You are not eligible for Vote")
except:
    print("Invalid input. Please enter the number value.")
```

### 4.2 Write a Python program using if-elif-else to calculate electricity charges based on the number of units consumed.

```python
total_bill_amount = 0

try:
    input_data = int(input("Please enter the units: "))
    
    if input_data > 1000:
        total_bill_amount = input_data * 11.55 
    elif input_data > 800 and input_data <= 1000:
        total_bill_amount = input_data * 10.5
    elif input_data > 600 and input_data <= 800:
        total_bill_amount = input_data * 9.45
    elif input_data > 500 and input_data <= 600:
        total_bill_amount = input_data * 8.4
    elif input_data > 400 and input_data <= 500:
        total_bill_amount = input_data * 6.3
    elif input_data > 100 and input_data <= 400:
        total_bill_amount = input_data * 4.7   
    else:
        total_bill_amount = 0

    print("Total Bill Amount is", total_bill_amount)
        
except:
    print("Invalid input. Please enter the number value.")
```

## 5. User Defined Functions without for loop, if loop

### 5.1 Write a Python program using if-elif-else to calculate electricity charges based on the number of units consumed. (Flat Calculation)

```python
def calculate_eb_bill(units):
    total_bill_amount = 0
    if units > 1000:
        total_bill_amount = units * 11.55 
    elif units > 800 and units <= 1000:
        total_bill_amount = units * 10.5
    elif units > 600 and units <= 800:
        total_bill_amount = units * 9.45
    elif units > 500 and units <= 600:
        total_bill_amount = units * 8.4
    elif units > 400 and units <= 500:
        total_bill_amount = units * 6.3
    elif units > 100 and units <= 400:
        total_bill_amount = units * 4.7   
    else:
        total_bill_amount = 0

    return total_bill_amount

try:
    input_data = int(input("Please enter the units: "))
    print("Total Bill Amount is", calculate_eb_bill(input_data))
except:
    print("Invalid input. Please enter the number value.")
```

## 6. User Defined Functions with for loop, if loop

### 6.1 Write a Python program using if-elif-else to calculate electricity charges based on the number of units consumed.

```python
def calculate_eb_bill(units):
    total_bill_amount = 0    
    while True:
        if units > 1000:
            print("units > 1000")
            total_bill_amount += ((units % 1000) * 11.55)
            units = 1000
        elif units > 800 and units <= 1000:
            print("units > 800 and units <= 1000")
            total_bill_amount += ((units % 800) * 10.5)
            if units == 1000:
                units = 800
            else:
                units = units - (units % 800)
        elif units > 600 and units <= 800:
            print("units > 600 and units <= 800")
            total_bill_amount += ((units % 600) * 9.45) 
            if units == 800:
                units = 600
            else:
                units = units - (units % 600)
        elif units > 500 and units <= 600:
            print("units > 500 and units <= 600")
            total_bill_amount += ((units % 500) * 8.4)
            if units == 600:
                units = 500
            else:
                units = units - (units % 500)
        elif units > 400 and units <= 500:
            print("units > 400 and input_data <= 500")
            total_bill_amount += ((units % 400) * 6.3)
            if units == 500:
                units = 400
            else:
                units = units - (units % 400)
        elif units > 100 and units <= 400:
            print("units > 100 and units <= 400")
            total_bill_amount += ((units % 100) * 4.7)
            if units == 400:
                units = 100
            else:
                units = units - (units % 100)
        else:
            return round(total_bill_amount, 2)
            break

        # print(units)

try:
    input_data = int(input("Please enter the units: "))
    print("Total Bill Amount is", calculate_eb_bill(input_data))
except:
    print("Invalid input. Please enter the number value.")
```

## 8. Break

### 8.1 Write a Python program using break to find the first repeated character in a given string.

```python
s_data = input("Please enter the string: ")

result = ""

for i in s_data:
    if i not in result or i == " ":
        result += i
    else:
        print("The first repeated character is -", i)
        break
        
```

### 8.2 Write a Python program using break to find the first number divisible by 7 in a given range.

```python
for i in range(1,10):
    if i % 7 == 0:
        break
    else:
        print(i)
```

## 9. Continue

### 9.1 Write a Python program using continue to print numbers from 1 to 10, skipping the number 5

```python
for i in range(1,10):
    if i == 5:
        continue
    else:
        print(i)
```

### 9.2 Write a Python program using continue to skip negative numbers in a list.

```python
list_nums = [12, 58, 89, -12, 78, -45, 100, -6]

results = []

for i in list_nums:
    if i > 0:
        results.append(i)
    else:
        continue

print("The new list are", results)
    
```
