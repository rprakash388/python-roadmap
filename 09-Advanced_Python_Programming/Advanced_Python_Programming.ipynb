```python
def set_customer_data(f_name, l_name, order_id, price):
    print("First Name:", f_name)
    print("Last Name:", l_name)
    print("Order Id:", order_id)
    print("Price:", price) 
```

```python
set_customer_data("Prakash", "Raju", 4332, 125.36)
```

```python
set_customer_data("Prakash", 1234, "Raju", 125.36)
```

### Positional Arguments

```python
set_customer_data(f_name="Prakash", order_id=1234, l_name="Raju", price=125.36)
```

### Variable length argument
single "*" means infinite arguments

```python
def set_customer_data1(f_name, l_name, *price):
    print("First Name:", f_name)
    print("Last Name:", l_name)
    print("Price:", price) 
```

```python
set_customer_data1("Prakash", "Raju", 45.7, 1589.65, 7000)
```

### Keyword Argument

```python
def set_customer_data2(f_name, l_name, **price):
    print("First Name:", f_name)
    print("Last Name:", l_name)
    print("Price:", price) 
```

```python
set_customer_data2("Prakash", "Raju", price_1=45.7, price_2=1589.65, price_3=7000)
```
