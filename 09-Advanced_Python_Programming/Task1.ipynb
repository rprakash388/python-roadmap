### 1.10 Format Map

 - works like format(), but it takes a dictionary as input instead of individual arguments.

```python
employee_data = {
    "name": "Prakash Raju",
    "age": 35
}

"Hello {name}, You are {age} years old.".format_map(employee_data)
```

```python
employee_data = {
    "name": "Prakash Raju",
    "age": 35,
    "salary": 56000
}

"Hello {name}, You are {age} years old. Salary: {salary}".format_map(employee_data)
```

## Use Case #2 – Count the number of repeated words in a sentence entered by the user.

```python
user_input2 = input("Please enter sentence: ")

words = user_input2.split()

results = {}

for word in words:    
    if word not in results:
        results[word] = 1
    else:
        results[word] += 1

for k,v in results.items():
    print(f"[{k}] count is {v}")
```
