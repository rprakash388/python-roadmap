## 3. Write a program that handles division by zero using try-except

```python
try:
    numerator = int(input("Enter Numerator: "))
    denaminator = int(input("Enter Denaminator: "))
    result = numerator/denaminator
    print(result)

except ZeroDivisionError:
    print("Please enter denaminator is greater than 0")

except ValueError:
    print("Please enter the numbers only")

finally:
    print("Program Ended !!!")
    
```

## 4. Give a string, count how many vowels it has

```python
user_input = input("Enter a string: ")

vowel_count = 0

for i in user_input.lower():
    if i in 'aeiou': 
        vowel_count += 1

print(f"Total number of vowels are: {vowel_count}")
        
```
