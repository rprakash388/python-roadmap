#### 1. Abstraction

- Abstraction means hiding implementation details and showing only essential features to the user.

```python
from abc import ABC, abstractmethod

class ATM(ABC):

    @abstractmethod
    def withdraw(self, pin, amount):
        pass


class SBIATM(ATM):

    def __init__(self):
        self.correct_pin = "1234"
        self.balance = 50000
        self.daily_limit = 20000
        self.atm_cash = 100000

    def validate_pin(self, pin):
        if pin != self.correct_pin:
            raise Exception("Invalid PIN")

    def check_account_status(self):
        print("✓ Account is Active")

    def check_daily_limit(self, amount):
        if amount > self.daily_limit:
            raise Exception("Daily withdrawal limit exceeded")

    def check_balance(self, amount):
        if amount > self.balance:
            raise Exception("Insufficient Balance")

    def check_atm_cash(self, amount):
        if amount > self.atm_cash:
            raise Exception("ATM does not have sufficient cash")

    def debit_account(self, amount):
        self.balance -= amount

    def update_bank_database(self):
        print("✓ Bank Database Updated")

    def dispense_cash(self, amount):
        self.atm_cash -= amount
        print(f"✓ Dispensing ₹{amount}")

    def print_receipt(self, amount):
        print("\n----- RECEIPT -----")
        print(f"Withdrawn Amount : ₹{amount}")
        print(f"Available Balance: ₹{self.balance}")
        print("-------------------")

    def log_transaction(self):
        print("✓ Transaction Logged")

    def withdraw(self, pin, amount):

        print("1. Validating PIN...")
        self.validate_pin(pin)

        print("2. Checking Account Status...")
        self.check_account_status()

        print("3. Checking Daily Withdrawal Limit...")
        self.check_daily_limit(amount)

        print("4. Checking Account Balance...")
        self.check_balance(amount)

        print("5. Checking ATM Cash Availability...")
        self.check_atm_cash(amount)

        print("6. Debiting Account...")
        self.debit_account(amount)

        print("7. Updating Bank Database...")
        self.update_bank_database()

        print("8. Dispensing Cash...")
        self.dispense_cash(amount)

        print("9. Printing Receipt...")
        self.print_receipt(amount)

        print("10. Logging Transaction...")
        self.log_transaction()

        print("\n✓ Transaction Successful")


atm = SBIATM()

try:
    atm.withdraw("1234", 10000)
except Exception as e:
    print("Transaction Failed:", e)
```

#### 2. Polymorphism

 - Polymorphism means "One Interface, Multiple Implementations".
 - The same method name can perform different actions depending on the object that calls it.

```python
class UPI:

    def pay(self, amount):
        print(f"₹{amount} paid using UPI")


class CreditCard:

    def pay(self, amount):
        print(f"₹{amount} paid using Credit Card")


class NetBanking:

    def pay(self, amount):
        print(f"₹{amount} paid using Net Banking")


payments = [
    UPI(),
    CreditCard(),
    NetBanking()
]

for payment in payments:
    payment.pay(5000)
```

##### Types of Polymorphism

1. Runtime Polymorphism (Method Overriding)
2. Compile-Time Polymorphism

##### 1. Runtime Polymorphism (Method Overriding)

 - Runtime Polymorphism occurs when a child class provides its own implementation of a method that already exists in the parent class, and the method to execute is decided at runtime.

```python
class Animal:
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print("Bark")

class Cat(Animal):
    def sound(self):
        print("Meow")

# Traditional way to call the method
dog = Dog() # Object creation step
dog.sound() # Method call step

cat = Cat()
cat.sound()

print("*-----------------------------*")
# Advanced way to call the method
sounds = [
    Dog(),
    Cat()
] 

for sound in sounds:
    sound.sound()
```

##### 2. Compile-Time Polymorphism

 - Compile-Time Polymorphism means the method to execute is determined at compile time, commonly through method overloading.
 - Python doesn't support traditional method overloading like Java/C#, but it can be achieved using default arguments.

```python
class Calculator:

    def add(self, a, b, c=0):
        return a + b + c

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

### 3. Inheritance

 - Inheritance allows one class to acquire properties and methods from another class.
 - Inheritance is an OOP concept where a child class inherits the properties and methods of a parent class, allowing code reuse and extension of existing functionality.

```python
class Employee: # Parent / Base Class

    def __init__(self, name):
        self.name = name

    def display(self):
        print("Employee:", self.name)

class Developer(Employee): # Child / Derived Class

    def code(self):
        print("Writing Python Code")

dev = Developer("Prakash")

dev.display()
dev.code()
```

##### Types of Inheritance in Python

1. Single Inheritance
2. Multiple Inheritance
3. Multi-Level Inheritance
4. Hierarchical Inheritance
5. Hybrid Inheritance
   

##### 1. Single Inheritance

 - One child class inherits from one parent class

```python
class Employee:
    def display(self):
        print("Employee")


class Developer(Employee):
    def code(self):
        print("Writing code")


dev = Developer()
dev.display()
dev.code()
```

##### 2. Multiple Inheritance

 - One child class inherits from multiple parent classes.

```python
class Developer:
    def code(self):
        print("Writing code")


class Tester:
    def test(self):
        print("Testing application")


class SoftwareEngineer(Developer, Tester):
    pass


employee = SoftwareEngineer()
employee.code()
employee.test()
```

##### 3. Multi-Level Inheritance

 - A child class inherits from a parent, and another class inherits from that child.

```python
class Employee:
    def work(self):
        print("Employee works")


class Developer(Employee):
    def code(self):
        print("Developer writes code")


class SeniorDeveloper(Developer):
    def design(self):
        print("Senior Developer designs application")


dev = SeniorDeveloper()
dev.work()
dev.code()
dev.design()
```

##### 4. Hierarchical Inheritance

 - Multiple child classes inherit from the same parent class.

```python
class Employee:
    def work(self):
        print("Employee works")


class Developer(Employee):
    def code(self):
        print("Developer writes code")


class Tester(Employee):
    def test(self):
        print("Tester tests application")


dev = Developer()
tester = Tester()

dev.work()
dev.code()

tester.work()
tester.test()
```

##### 5. Hybrid Inheritance

 - A combination of two or more types of inheritance

```python
class Employee:
    def __init__(self, name, employee_id):
        self.name = name
        self.employee_id = employee_id

    def work(self):
        print(f"{self.name} is working as an employee.")


class Developer(Employee):
    def code(self):
        print(f"{self.name} is writing code.")


class Tester(Employee):
    def test(self):
        print(f"{self.name} is testing the application.")


class TechLead(Developer, Tester):
    def lead(self):
        print(f"{self.name} is managing the development team.")

    def display_details(self):
        print("----- Employee Details -----")
        print("Name       :", self.name)
        print("Employee ID:", self.employee_id)


# Create TechLead object
tech_lead = TechLead("Prakash", "EMP101")

# Inherited from Employee
tech_lead.work()

# Inherited from Developer
tech_lead.code()

# Inherited from Tester
tech_lead.test()

# TechLead's own methods
tech_lead.lead()
tech_lead.display_details()
```
